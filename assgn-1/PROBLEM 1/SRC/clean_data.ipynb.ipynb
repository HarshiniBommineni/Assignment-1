```python
import pandas as pd
import os

# Load the CSV file
df = pd.read_csv('/content/PROBLEM1.csv')

# Strip column names to remove any extra spaces
df.columns = df.columns.str.strip()

# Drop rows with missing critical values
df = df.dropna(subset=["Age", "Height", "Weight", "Frailty"])

# Fill missing Grip strength with median (if the column exists)
if "Grip strength" in df.columns:
    df["Grip strength"] = df["Grip strength"].fillna(df["Grip strength"].median())

# Save the cleaned data to a new CSV
os.makedirs('/content/cleaned_data', exist_ok=True)
clean_file_path = '/content/cleaned_data/PROBLEM1_clean.csv'
df.to_csv(clean_file_path, index=False)

print(f"Cleaned data saved to: {clean_file_path}")
print(df)
```
