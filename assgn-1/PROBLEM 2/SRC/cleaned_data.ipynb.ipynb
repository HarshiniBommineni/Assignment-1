```python
import pandas as pd
import os

# Load the dataset
df = pd.read_csv('/content/StudentsPerformance.csv')

# Strip column names to remove extra spaces
df.columns = df.columns.str.strip()

# Drop rows with missing critical values
df = df.dropna()  # drops any row with at least one missing value

# Optional: fill numeric columns missing values with median
numeric_cols = df.select_dtypes(include=['number']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Optional: fill categorical columns missing values with mode
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Save the clean data
os.makedirs('/content/cleaned_data', exist_ok=True)
clean_file_path = '/content/cleaned_data/student_performance_clean.csv'
df.to_csv(clean_file_path, index=False)

print(f"Clean data saved to: {clean_file_path}")
print(df.head())
```
